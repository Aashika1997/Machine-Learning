```python
import pandas as pd
dataset=pd.read_csv("Social_Network_Ads.csv")
dataset
dataset=pd.get_dummies(dataset,drop_first=True)
dataset=dataset.drop(columns=["User ID"])
dataset
x=dataset[["Age","EstimatedSalary","Gender_Male"]]
y=dataset[["Purchased"]]
print(y)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=1/8,random_state=52)
dataset[["Purchased"]].value_counts()
dataset.shape
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import ComplementNB
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, r2_score

pipe = Pipeline([
    ('scaler', MinMaxScaler()),
    ('classifier', ComplementNB())
])

param_grid = {
    'classifier__alpha': [0.0001, 0.001, 0.01, 0.1, 0.5, 1, 2, 5, 10],
    'classifier__fit_prior': [True, False],
    'classifier__norm': [True, False]
}

grid = GridSearchCV(
    pipe,
    param_grid,
    cv=5,
    scoring='accuracy',
    refit=True,
    verbose=3,
    n_jobs=-1
)

grid.fit(x_train, y_train.values.ravel())

print("Best Parameters:", grid.best_params_)
print("Best Accuracy:", grid.best_score_)

y_pred = grid.best_estimator_.predict(x_test)

print("Test Accuracy:", accuracy_score(y_test, y_pred))
print("R_score:", r2_score(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
re=grid.cv_results_ 
grid_predictions = grid.predict(x_test)
from sklearn.metrics import r2_score
r_score=r2_score(y_test,grid_predictions)
print("The R_score value for best parameter {}:".format(grid.best_params_),r_score)
```
